```python
import numpy as np

# Create 1-D NumPy array with 25 evenly spaced numbers from 10 to 100
paryag = np.linspace(10, 100, 25)

print("Array:", paryag)

# Print dimensions
print("Dimensions:", paryag.ndim)

# Print shape
print("Shape:", paryag.shape)

# Print total number of elements
print("Total Elements:", paryag.size)

# Print data type
print("Data Type:", paryag.dtype)

# Print total number of bytes
print("Total Bytes:", paryag.nbytes)

# Find transpose using reshape()
transpose = paryag.reshape(1, 25).T

print("Transpose using reshape():")
print(transpose)

# Find transpose using T attribute
print("Transpose using T:")
print(paryag.T)
```

```python
import numpy as np

# Create a 2-D array with 3 rows and 4 columns
ucs420_paryag = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:")
print(ucs420_paryag)

# Mean
print("Mean:", np.mean(ucs420_paryag))

# Median
print("Median:", np.median(ucs420_paryag))

# Maximum
print("Maximum:", np.max(ucs420_paryag))

# Minimum
print("Minimum:", np.min(ucs420_paryag))

# Unique elements
print("Unique Elements:", np.unique(ucs420_paryag))

# Reshape to 4 rows and 3 columns
reshaped_ucs420_paryag = ucs420_paryag.reshape(4, 3)

print("Reshaped Array (4 x 3):")
print(reshaped_ucs420_paryag)

# Resize to 2 rows and 3 columns
resized_ucs420_paryag = np.resize(ucs420_paryag, (2, 3))

print("Resized Array (2 x 3):")
print(resized_ucs420_paryag)
```
